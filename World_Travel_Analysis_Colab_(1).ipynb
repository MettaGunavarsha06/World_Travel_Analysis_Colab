## Part 4 — Visualise the results

```python
import matplotlib.pyplot as plt
continent_counts = df['Continent'].value_counts()
continent_counts.plot(kind='bar', title='Travel Profiles by Continent')
plt.xlabel('Continent')
plt.ylabel('Number of profiles')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
trip_summary.plot(kind='bar', title='Average Estimated Trip Budget by Trip Type')
plt.xlabel('Trip type')
plt.ylabel('Average trip budget (USD)')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
df['Estimated_Daily_Budget_USD'].plot(kind='hist', bins=20, title='Daily Budget Distribution')
plt.xlabel('Estimated daily budget (USD)')
plt.tight_layout()
plt.show()
```

```python
plt.scatter(df['Estimated_Daily_Budget_USD'], df['Average_Traveler_Rating'], alpha=0.5)
plt.title('Estimated Daily Budget vs Sample Rating')
plt.xlabel('Estimated daily budget (USD)')
plt.ylabel('Sample rating')
plt.tight_layout()
plt.show()
```

```python
season_counts = df['Best_Season'].value_counts()
season_counts.plot(kind='pie', autopct='%1.1f%%', title='Best Season Distribution')
plt.ylabel('')
plt.show()
```

## Part 5 — More analysis

```python
corr = df[['Recommended_Days', 'Estimated_Daily_Budget_USD', 'Estimated_Trip_Budget_USD', 'Average_Traveler_Rating']].corr()
print(corr.round(2))
```

```python
corr = df[['Recommended_Days', 'Estimated_Daily_Budget_USD', 'Estimated_Trip_Budget_USD', 'Average_Traveler_Rating']].corr()
plt.imshow(corr, aspect='auto')
plt.xticks(range(len(corr.columns)), corr.columns, rotation=45, ha='right')
plt.yticks(range(len(corr.columns)), corr.columns)
plt.colorbar(label='Correlation')
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

```python
city_avg = df.groupby('Destination')['Estimated_Trip_Budget_USD'].mean().sort_values()
print('10 lowest estimated average trip budgets:')
print(city_avg.head(10))
```

```python
print('10 highest estimated average trip budgets:')
print(city_avg.tail(10).sort_values(ascending=False))
```
